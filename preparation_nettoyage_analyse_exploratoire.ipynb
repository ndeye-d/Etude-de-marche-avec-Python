```python
#On garde les colonnes utiles
df_population = df_population_init[['Zone', 'Élément','Valeur']]
df_population.head()
```

```python
#Multiplication des populations par 1000
df_population.loc[:, 'Valeur'] = df_population.loc[:,'Valeur'] * 1000
df_population.head()
```

```python
#Pivotage du dataframe
df_population = df_population.pivot(index='Zone', columns='Élément', values='Valeur').reset_index()
df_population.columns.name=None
df_population.head()
```

```python
#Renommage de la colonne Zone
df_population = df_population.rename(columns={'Zone': 'Pays'})
df_population.head()
```

```python
print("Statistiques descriptive du dataset population")
display(df_population.describe())
```

```python
#Recherche de valeurs vide
print(df_population.isnull().sum())
```

```python
#Ligne contenant des valeurs nulles
df_population[df_population['Population totale'].isnull()]
```

```python
#Calcul de la population total des pays Saint-Siège et Îles Anglo-Normandes	
mask_saint_siege = df_population['Pays']== 'Saint-Siège'
df_population.loc[mask_saint_siege, 'Population totale'] = df_population.loc[mask_saint_siege, 'Population rurale'] + df_population.loc[mask_saint_siege, 'Population urbaine']

mask_iles_anglo_normandes = df_population['Pays']== 'Îles Anglo-Normandes'
df_population.loc[mask_iles_anglo_normandes, 'Population totale'] = df_population.loc[mask_iles_anglo_normandes, 'Population rurale'] + df_population.loc[mask_iles_anglo_normandes, 'Population urbaine']
```

```python
#Ligne contenant des valeurs nulles
df_population[df_population['Population urbaine'].isnull()]
```

```python
#Ligne contenant des valeurs nulles
df_population[df_population['Population rurale'].isnull()]
```

```python
#Recherche de pays dupliqué
df_population.loc[df_population["Pays"].duplicated(keep=False),:]
```

```python
#Ajout d'une colonne contenant la proportion de population urbaine
df_population.loc[:, 'Pourcentage de population urbaine'] = df_population.loc[:, "Population urbaine"] / df_population.loc[:, "Population totale"] * 100

#Suppression des colonnes Population rurale et Population urbaine
df_population = df_population.drop(labels=['Population rurale', "Population urbaine"], axis=1)

df_population.head()
```

```python
#Suppression de la Chine car sa population est las somme des population de la Chine continentale, de Macao, de Hong-KOng et de Taiwan
df_population = df_population.drop(df_population[df_population["Pays"] == "Chine"].index).reset_index(drop=True)
```
